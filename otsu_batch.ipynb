**[Task 2.1.]()** Implement Otsu thresholding as a re-usable function:

```python
def otsu_thresholding(imgf):
    img8 = (imgf * 255).round().astype(np.uint8)
    list_T=[]
    for t in range (1,256):
        class1=img8[img8<t]
        class2=img8[img8>=t]
        if class1.size==0 or class2.size==0:
            T=np.inf
        else:
            T=class1.size*np.var(class1)+class2.size*np.var(class2)
    
        list_T.append(T)
        seg =list_T.index(min(list_T))+1
    print("optimal threshold:", seg)
    
  
    
    return seg
```

**[Task 2.2.]()** Implement a re-usable function to compute the Dice coefficient:

```python
def compute_dice(bin1, bin2):
    threshold = otsu_thresholding(bin2)
    bin1 = bin1.astype(bool)
    bin2 = bin2.astype(bool)
    assert bin1.dtype == bool
    assert bin2.dtype == bool

    H = np.asarray(bin1>0, dtype=bool)
    G = np.asarray(bin2>=threshold/255, dtype=bool)
    GH = np.logical_and(G, H).sum()
    dice = 2 * GH / (G.sum() + H.sum())

    return dice
```
