**[Task 2.2.]()** Implement the two functions `sobel_h` and `sobel_v` for horizontal and vertical $3 \times 3$ Sobel filtering, respectively:

```python

def sobel_h(img):
    kernelx= np.array([[-1,0,1],[-2,0,2],[-1,0,1]])
    result= np.zeros(img.shape)
    for y in range (1,img.shape[0]-1):
        for x in range (1,img.shape[1]-1):
            region= img[y-1:y+2,x-1:x+2]
            result[y,x]= (region*kernelx).sum()
    return result

def sobel_v(img):
    kernely= np.array([[-1,-2,-1],[0,0,0],[1,2,1]])
    result=np.zeros(img.shape)
    for y in range (1,img.shape[0]-1):
            for x in range (1,img.shape[1]-1):
                region= img[y-1:y+2,x-1:x+2]
                result[y,x]= (region*kernely).sum()
    return result
  
```
